# vizly Jupyter gallery

English-first trust defaults (local/allowlisted assets, `en-US` chrome) with worldwide map atlas by default.

```python
import pandas as pd
import vizly as vz

vz.set_theme("corporate")
df = pd.DataFrame({
    "date": pd.to_datetime(["2026-01-01", "2026-01-02", "2026-01-03"]),
    "revenue": [12, 18, 15],
    "cost": [5, 7, 6],
})
vz.line(df, x="date", y=["revenue", "cost"], title="Revenue vs cost")
```

```python
vz.pie({"name": ["East", "West", "Central"], "value": [40, 35, 25]}, title="Share")
```

```python
vz.map(
    pd.DataFrame({"name": ["United States", "Canada", "Mexico"], "value": [30, 18, 12]}),
    title="World sample",
)
```
